# Step 3: Data Preprocessing

## Objectives
- Load and validate the original dataset.
- Separate input features and target variable.
- Inspect feature variance and data quality.
- Split data into training and testing sets.
- Prepare preprocessing steps for machine learning.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/phishing_dataset.csv")

print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

display(df.head())
```

```python
# Input features
X = df.drop(columns=["url", "label"])

# Target variable
y = df["label"]

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget classes:")
print(y.value_counts())
```

```python
feature_variance = X.nunique(dropna=False).sort_values()

print("Number of unique values in each feature:")
display(feature_variance.to_frame("Unique Values"))

constant_features = feature_variance[
    feature_variance <= 1
].index.tolist()

print("Constant features:", constant_features)
```

```python
# Remove constant features from the input data
X = X.drop(columns=constant_features)

print("Features after removing constant columns:", X.shape)
print("Remaining constant features:", X.columns[X.nunique() <= 1].tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
display(y_train.value_counts())

print("\nTesting target distribution:")
display(y_test.value_counts())
```

```python
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print(
    "Overlap in original row indices:",
    len(X_train.index.intersection(X_test.index))
)
```
